# ระบบสั่งอาหาร (Food Ordering System)

## รายละเอียดโครงงาน

โครงงานนี้เป็นระบบ POS (Point of Sale) สำหรับสั่งอาหารอัตโนมัติ พัฒนาด้วยภาษา Python บน Google Colab เพื่อช่วยในการจัดการการขายหน้าร้าน ตั้งแต่การรับออเดอร์ คำนวณราคา รับชำระเงิน ออกใบเสร็จ และสรุปยอดขายทั้งหมด

## ความสามารถของระบบ

User
- ลูกค้าสามารถกดสั่งอาหารได้
- เลือกท็อปปี้ได้ตามใจ
- เลือกเมนูตามที่ AI แนะนำได้
- สามารถเพิ่มลงตะกร้าได้
- สามารถแก้ไขจำนวนหรือลบเมนูที่ไม่ต้องการได้
- สั่งอาหารได้หลายรายการ
- ยืนยันการสั่งอาหาร
- คำนวณราคารวมอัตโนมัติ
- ดูรายละเอียดการสั่งซื้อ
- กดเรียกชำระเงิน
- ดูสถานะการชำระเงิน


Admin
- เข้าสู่ระบบ
- สามารถดูออเดอร์ปัจจุบันได้
- สามารถดูสถานะการชำระเงินได้
- สามารถพิมพ์ใบเสร็จได้
- จัดการอาหารหลัก เพิ่ม ลบ แก้ไขได้
- จัดการท็อปปี้ เพิ่ม ลบ แก้ไขได้
- สรุปยอดขายประจำวัน/ประจำเดือน/ประจำปี

## วิธีใช้งาน

1. เปิดโปรเจ็ค : เปิดไฟล์ Food_Ordering_System.ipynb ด้วย Google Colab

2. เริ่มต้นระบบ : รันเซลล์ทั้งหมดตามลำดับ หรือกดเมนู Runtime > Run all

3. ฝั่งลูกค้า (Customer Interface):
- กรอกชื่อลูกค้าหรือหมายเลขโต๊ะ
- เลือกเมนูอาหาร หรือเครื่องดื่ม/ของหวาน พร้อมระบุจำนวน
- เลือกท็อปปิ้ง และปรับระดับความเผ็ด/ความหวานตามต้องการ (หรือกดใช้ AI จับคู่เมนูสมาร์ต)
- กดปุ่มเพิ่มลงตะกร้า เพื่อตรวจเช็กรายการ
- กดปุ่มแก้ไขจำนวนหรือลบเมนูที่ไม่ต้องการได้
- กดปุ่มยืนยันการสั่ง (ส่งเข้าครัว)เพื่อส่งออเดอร์ไปยังระบบหลังบ้าน และระบบจะคำนวณราคาที่ต้องชำระทั้งหมดรวม vat

4. ฝั่งแอดมิน / แคชเชียร์ (Admin POS Center):
- ตรวจสอบออเดอร์เข้าใหม่บน Live Orders Dashboard
- เลือกช่องทางชำระเงิน (เงินสด หรือ QR PromptPay) และกรอกยอดเงินที่รับมา
- กดปุ่ม พิมพ์สลิป & ปิดออเดอร์ เพื่อออกใบเสร็จรับเงิน

5. ติดตามสถานะเรียลไทม์ (Real-time Sync):
- หน้าฝั่งลูกค้าจะอัปเดตสถานะเป็น "ชำระเงินเสร็จสิ้นแล้ว" เมื่อแอดมินกดพิมพ์สลิป & ปิดออเดอร์
- หน้าฝั่งแอดมินจะอัปเดตสถานะเป็น "ลูกค้าเรียกชำระเงิน" เมื่อลูกค้ากดปุ่มชำระเงิน

6. จัดการหลังบ้าน & สรุปยอดขาย (Back-Office & Analytics):
- จัดการรายการเมนู เพิ่ม/แก้ไขราคา/ลบรายการ ได้ในส่วน Admin Panel
- เรียกดูรายงานสรุปยอดขายรายวัน รายเดือน รายปี
