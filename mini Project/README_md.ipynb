# 🍽️ ระบบสั่งอาหาร (Food Ordering System)

## รายละเอียดโครงงาน
โครงงานนี้เป็นระบบ POS (Point of Sale) สำหรับสั่งอาหารและออกใบเสร็จอัตโนมัติ พัฒนาด้วยภาษา Python ช่วยในการจัดการการขายหน้าร้าน ตั้งแต่การรับออเดอร์ เลือกระดับความเผ็ด/หวาน เลือกท็อปปิ้ง ค้นหาเมนู คำนวณราคารวม ส่วนลด ภาษีมูลค่าเพิ่ม (VAT 7%) พิมพ์ใบเสร็จรับเงิน และบันทึกประวัติออเดอร์ลงไฟล์ JSON อัตโนมัติ

## วัตถุประสงค์
- เพื่อเพิ่มความสะดวกและรวดเร็วในการรับออเดอร์หน้าร้าน (POS)
- เพื่อคำนวณราคาอาหาร ส่วนลดโปรโมชั่น และภาษี VAT อย่างแม่นยำ
- เพื่อค้นหาและเลือกท็อปปิ้ง/ระดับความหวานหรือความเผ็ดตามความต้องการของลูกค้า
- เพื่อจัดการข้อมูลสินค้าในตะกร้า (เพิ่ม ลบรายการ จัดเรียงลำดับราคา)
- เพื่อบันทึกประวัติรายการสั่งซื้อลงไฟล์ข้อมูล JSON อัตโนมัติ

## ความสามารถของระบบ

### ฝั่งผู้ใช้งาน / ลูกค้า (User)
- ระบุชื่อลูกค้าหรือหมายเลขโต๊ะได้
- ดูรายการอาหาร แบ่งตามหมวดหมู่ (อาหารคาว / เครื่องดื่ม)
- ค้นหาเมนูอาหารตามชื่อ (Search Feature)
- เลือกระดับความหวาน / ความเผ็ด (เช่น ปกติ, หวานน้อย, เผ็ดมาก)
- เลือกท็อปปิ้งเสริม (เช่น ไข่ดาว, ชีส, ไข่มุก)
- เพิ่มรายการสินค้าลงตะกร้าพร้อมระบุจำนวน
- ตรวจสอบรายการในตะกร้า คำนวณราคาสุทธิ ส่วนลด และ VAT เบื้องต้น
- ลบรายการที่ไม่ต้องการออกจากตะกร้าตามลำดับ
- จัดเรียงรายการสินค้าในตะกร้าตามราคาสูง-ต่ำ
- ชำระเงิน สรุปยอด และพิมพ์ใบเสร็จรับเงิน (Receipt) พร้อมลงประทับเวลา
- บันทึกประวัติคำสั่งซื้อลงไฟล์ `orders.json` อัตโนมัติ

## วิธีใช้งาน

1. **เริ่มต้นระบบ:** รันไฟล์โปรแกรม `Sourcecode.py` ผ่าน Python CLI, Terminal หรือ Google Colab
2. **กรอกข้อมูลลูกค้า:** ใส่ชื่อลูกค้าหรือหมายเลขโต๊ะ (หากเว้นว่างจะตั้งเป็น "ลูกค้าทั่วไป")
3. **เลือกทำรายการจากเมนูหลัก (1-5):**
   - **กด 1 [ดูเมนูอาหาร & ค้นหาเมนู]:** แสดงรายการอาหาร ท็อปปิ้ง และพิมพ์ชื่อเมนูเพื่อค้นหา
   - **กด 2 [เลือกสั่งสินค้าลงตะกร้า]:** เลือกรหัสเมนู ระบุจำนวน ปรับระดับความเผ็ด/หวาน และเลือกรหัสท็อปปิ้งเสริม
   - **กด 3 [ดูตะกร้า / ลบรายการ / เรียงลำดับราคา]:**
     - กด `A` เพื่อเลือก ลบรายการสินค้าออกจากตะกร้า
     - กด `B` เพื่อ เรียงลำดับสินค้าในตะกร้าตามราคาสูง-ต่ำ
   - **กด 4 [ชำระเงิน & พิมพ์ใบเสร็จ]:** คำนวณส่วนลด 10% (เมื่อสั่งครบ 300 บาท) + VAT 7% พิมพ์สลิปใบเสร็จ และบันทึกข้อมูลลง `orders.json`
   - **กด 5 [ออกจากระบบ]:** จบการทำงานของโปรแกรม
