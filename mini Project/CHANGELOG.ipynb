# บันทึกการเปลี่ยนแปลง (CHANGELOG)

## วันที่ 1

### เพิ่ม (Added)
- **โครงสร้างโปรแกรมหลัก**

- ออกแบบลูปการทำงานหลักของโปรแกรม (main()) รองรับเมนูโต้ตอบแบบ Command Line (CLI)

- เพิ่มระบบระบุชื่อลูกค้า / เลขโต๊ะ (customer_name)
- **ระบบจัดการข้อมูลเมนู**

- กำหนดชุดข้อมูลเมนูอาหาร เครื่องดื่ม และท็อปปิ้งเริ่มต้น (default_menu, default_toppings)

- พัฒนาฟังก์ชันแสดงรายการเมนูอาหารและเครื่องดื่ม (display_menu())

- พัฒนาฟังก์ชันแสดงรายการท็อปปิ้งเสริม (display_toppings())
- **ระบบค้นหาเมนู**

- พัฒนาระบบค้นหาเมนูอาหารจากคำค้นหา (Case-insensitive search)
- **ระบบรับออเดอร์ลงตะกร้า**

- พัฒนาระบบเลือกสั่งสินค้า ระบุจำนวน ระดับความหวาน/ความเผ็ด และการเลือกท็อปปิ้ง

- คำนวณราคาต่อหน่วยและราคารวมของแต่ละรายการ บันทึกลงตะกร้าสินค้า (cart.append())

## วันที่ 2

### เพิ่ม (Added)
- **ระบบจัดการตะกร้าสินค้าขั้นสูง**

- เพิ่มฟังก์ชันลบรายการสินค้าออกจากตะกร้าตามลำดับที่ต้องการ (cart.pop())

- เพิ่มฟังก์ชันเรียงลำดับรายการสินค้าในตะกร้าตามราคาสูง-ต่ำ (cart.sort())

- **ระบบคำนวณเงินและชำระเงิน**

- พัฒนาฟังก์ชัน calculate_bill() คำนวณส่วนลด 10% เมื่อยอดสั่งซื้อครบ 300 บาท และคำนวณภาษี VAT 7%

- พัฒนาระบบพิมพ์ใบเสร็จรับเงิน (Receipt) พร้อมประทับวันที่และเวลาปัจจุบันด้วย datetime

- **ระบบจัดเก็บข้อมูลลงไฟล์**

- พัฒนาฟังก์ชัน load_json() และ save_json() สำหรับอ่าน/บันทึกข้อมูลเมนู ท็อปปิ้ง และประวัติคำสั่งซื้อ (orders.json)

- เพิ่มระบบแปลง Key ของ Dictionary จาก String กลับเป็น int อัตโนมัติเมื่อดึงข้อมูลจากไฟล์ JSON

- เพิ่มระบบล้างตะกร้า (cart.clear()) เพื่อรีเซ็ตระบบสำหรับลูกค้ารายถัดไป
