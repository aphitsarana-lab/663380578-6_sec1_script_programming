# วันที่ 2 : พัฒนาระบบและทดสอบ

## งานที่ทำ

* พัฒนาระบบเลือกสั่งสินค้า พร้อมตัวเลือกเพิ่มเติม (ระดับความหวาน/ความเผ็ด, ท็อปปิ้ง)

* พัฒนาฟังก์ชันคำนวณยอดรวม (calculate_bill) พร้อมเงื่อนไขส่วนลด 10% เมื่อยอดซื้อถึง 300 บาท และคำนวณ VAT 7%

* พัฒนาระบบจัดการตะกร้าสินค้า: แสดงรายการ, ลบรายการสินค้าตามลำดับ (pop), และเรียงลำดับสินค้าตามราคา (sort)
* พัฒนาระบบสรุปรายการและพิมพ์ใบเสร็จรับเงิน (Receipt System)

* ดึงเวลาปัจจุบันด้วยโมดูล datetime เพื่อลงบันทึกในใบเสร็จ

* บันทึกประวัติคำสั่งซื้อสะสมลงไฟล์ orders.json

* ระบบรีเซ็ต/ล้างตะกร้าสินค้า (cart.clear()) เพื่อรองรับลูกค้ารายถัดไป

## การใช้ AI

**จำนวน Prompt ที่ใช้:** ประมาณ 5 Prompt

ตัวอย่าง Prompt

* วิธีคำนวณส่วนลดตามเงื่อนไขและคิดภาษี VAT 7% ด้วย Python

* วิธีการลบ Element ใน List ตาม Index ที่ผู้ใช้เลือกแบบปลอดภัย

* วิธีเรียงลำดับ List of Dictionaries ตาม value ที่ต้องการด้วย lambda
* วิธีดึงเวลาปัจจุบันและจัดฟอร์แมตวันเวลาด้วย datetime.now().strftime"

* วิธีล้างข้อมูลทั้งหมดใน List เพื่อเริ่มต้นใหม่และการแสดงผลตัวเลขทศนิยม 2 ตำแหน่งด้วย f-string ใน Python"

## สิ่งที่ได้เรียนรู้

* การประยุกต์ใช้ List Operations สำคัญ เช่น .append(), .pop(), .sort(), และ len()

* การใช้ Lambda Function เพื่อกำหนดเงื่อนไขการเรียงลำดับ (key=lambda x: x['total_price'])

* การสร้าง Business Logic สำหรับคำนวณเงิน (Subtotal, Discount, VAT, Grand Total)
* การทำงานกับโมดูล datetime และการจัดฟอร์แมตวันที่

* การจัดรูปแบบข้อความและตัวเลขด้วย f-string (เช่น :.2f สำหรับแสดงทศนิยม 2 ตำแหน่ง)

* การจัดระเบียบ Workflow ของโปรแกรมให้รองรับการทำงานต่อเนื่อง (Main Loop)

## ปัญหาที่พบ

* ผู้ใช้กรอกลำดับรายการที่ต้องการลบเกินจำนวนที่มีในตะกร้า ทำให้เกิด Error (IndexError)
* ผู้ใช้กดชำระเงินทั้งๆ ที่ยังไม่มีสินค้าในตะกร้า

## วิธีแก้ไข

* เพิ่มการตรวจสอบขอบเขตข้อมูล (1 <= remove_idx <= len(cart)) ก่อนสั่งลบข้อมูลด้วย .pop()
* เพิ่มการเช็กเงื่อนไข if not cart: เพื่อแจ้งเตือนและตัดการทำงานก่อนคำนวณใบเสร็จ

# สรุปการเรียนรู้

* ด้านโครงสร้างโปรแกรม: เข้าใจการแบ่งหน้าที่โปรแกรมออกเป็นฟังก์ชันย่อย (Modular Programming) ช่วยให้โค้ดอ่านง่ายและแก้ไขได้สะดวก

* ด้านการจัดการข้อมูล: ได้ฝึกทักษะการควบคุมข้อมูลในหน่วยความจำชั่วคราว (List/Dict) และการจัดเก็บลงพื้นที่ถาวร (JSON Persistence)

* ด้าน Error Handling: ได้เรียนรู้การป้องกันโปรแกรมค้าง/ล่มด้วย try-except และการ Validate ข้อมูลนำเข้าของผู้ใช้ (Input Validation)

* ด้านการใช้ AI: สามารถใช้ AI ในการสืบค้นไวยากรณ์ Python ที่ซับซ้อน (เช่น Lambda, File Handling) และให้ AI ช่วยเสนอโครงสร้างโค้ดที่รัดกุมขึ้น
